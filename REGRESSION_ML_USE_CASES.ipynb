    - O Erro Médio Quadrático (EMQ) — ou Mean Squared Error (MSE), em inglês — é uma métrica que serve para medir a qualidade de um modelo de previsão ou estimação (como regressões lineares, IA e modelos estatísticos).
```python
# Installing required libraries

import numpy as np
from pydataset import data

# from sklearn.preprocessing import LabelEncoder
from sklearn import preprocessing
from sklearn import linear_model
from sklearn.model_selection import KFold
```

```python
# Get the Boston dataset from the pydataset library
dataset = data('Boston')
```

```python
# Verifying the first few rows of the dataset
dataset.head()
```

```python
dataset.describe()
```

```python
# Removing outliers from the target variable 'medv'
dataset = dataset[~(dataset['medv'] >= 50.0)]  

prices = dataset['medv']
features = dataset.drop('medv', axis=1)

info = "The dataset has been successfully loaded and preprocessed. \nThe target variable 'medv' has been cleaned of outliers,\nand the features have been separated for further analysis."

print("$"*80)
print(f"\n{info}\n")
print("$"*80)

print("\n")
print("Features shape (lines, columns) :  ", features.shape)
print("Prices shape (lines, columns):   ", prices.shape)  

# print("BOSTON HOUSING DATASET HAS {} DATA POINTS WITH {} VARIABLES EACH.".format(dataset.shape[0], dataset.shape[1]))

print("BOSTON HOUSING DATASET HAS < {} > DATA POINTS WITH < {} > VARIABLES EACH.".format(*dataset.shape))
```

```python
# Get the minimum and maximum prices from the dataset

minimum_price = np.amin(prices)

maximum_price = np.amax(prices)

mean_price = np.mean(prices)

median_price = np.median(prices)

std_price = np.std(prices)

#  Show the calculated statistics for the target variable 'medv'

print("\n")
print("Statistics for Boston housing dataset over the target variable 'medv':")
print("\n")

print("Minimum price: ${:,.2f}".format(minimum_price))
print("Maximum price: ${:,.2f}".format(maximum_price))
print("Mean price: ${:,.2f}".format(mean_price))
print("Median price ${:,.2f}".format(median_price))
print("Standard deviation of prices: ${:,.2f}".format(std_price))
```

```python
#  Selecting the variables to be used for the regression model, based on their correlation with the target variable 'medv'

selected_variables1 = [
    "INDUS", "NOX", "RM", "AGE", "DIS", "TAX", "PTRATIO", "LSTAT"
]

selected_variables2 = [
    "CRIM", "ZN", "INDUS", "CHAS", "RAD", "BLACK", "LSTAT", "TAX"
]

# create a random selection to choose between the two sets of selected variables for the regression model
#  [1, 3[ mean generates a random integer between 1 and 2 (inclusive of 1, exclusive of 3)]
ran = np.random.randint(1, 3)

match ran:
    case 1:
        print("\n")
        print("Selected set 1 of variables for regression model: \n\n", selected_variables1)
        print("\n")
        
        features = dataset[[v.lower() for v in selected_variables1]]
    case 2:
        print("\n")
        print("Selected set 2 of variables for regression model: \n\n", selected_variables2)
        print("\n")
        features = dataset[[v.lower() for v in selected_variables2]]
```

##### TRAIN AND TEST MODEL LinearRegrassion

```python
# Preparing the model for training and testing
#  MODEL: Linear Regression, with 10-fold cross-validation and Min-Max scaling of features

# ---------------------------------------------------------------


l_regression = linear_model.LinearRegression()

# Definindo o número de folds para validação cruzada
kf = KFold(n_splits=10)

# Normalizing the features using Min-Max scaling, which scales the features to a range between 0 and 1. This is important for many machine learning algorithms that are sensitive to the scale of input data.
min_max_scaler = preprocessing.MinMaxScaler()
x_scaled = min_max_scaler.fit_transform(features)

# min_max_scaler2 = preprocessing.MinMaxScaler()
# x_scaled2 = min_max_scaler2.fit_transform(features_two)


print("\n")
print(f"X_scaled : \n, {x_scaled[1]}")
# print("\n")
# print(f"X_scaled2 : \n, {x_scaled2[1]}")
# print("\n")
```

##### CROSS VALIDATION AND RESULTS OF ML MODEL LinearRegression

```python
from sklearn.model_selection import cross_val_score

lr_scores = cross_val_score(l_regression, x_scaled, prices, cv=kf, scoring='neg_mean_absolute_error')

print("="*80)
print("RESULTS FOR LINEAR REGRESSION MODEL  (negative mean absolute error) ")
print("="*80)

print("\n")

print(f" ::>> ALL Cross-validation scores: {lr_scores}")
print(" ::>> MSE: {:.2f}".format(lr_scores.mean()))
print(" ::>> Standard deviation:  {:.2f}".format(lr_scores.std()))
```

##### TRAIN AND TEST MODEL RandomForestClassifier

```python
# Preparing the model for training and testing
# MODEL: Random Forest Regressor, which is an ensemble learning method that operates by constructing multiple decision trees during training and outputting the mean prediction of the individual trees. It is used for regression tasks and can handle both numerical and categorical data.

# -----------------------------------------------------------------------------

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score

regressor_rf = RandomForestRegressor(n_estimators=100, random_state=0) 

# Definindo o número de folds para validação cruzada
# kf = KFold(n_splits=10)

# Training the model using cross-validation and calculating the MSE scores
rf_scores = cross_val_score(regressor_rf, x_scaled, prices, cv=kf, scoring='neg_mean_absolute_error')
 

print("="*80)
print("RESULTS FOR RANDOM FOREST REGRESSOR MODEL  (negative mean absolute error) ")
print("="*80)

print(f" ::>> ALL Cross-validation scores: {rf_scores}")
print(f" ::>> MSE: {round(rf_scores.mean(), 2)}")
print(f" ::>> Standard deviation: {round(rf_scores.std(), 2)}")
```

##### TRAIN AND TEST MODEL GradientBoostingClassifier

```python
# Preparing the model for training and testing
# MODEL: Gradient Boosting Regressor, which is an ensemble learning method that builds a predictive model in a stage-wise fashion. It is used for regression tasks and can handle both numerical and categorical data. It combines the predictions of several base estimators to improve robustness over a single estimator.

# -----------------------------------------------------------------------

from sklearn.ensemble import GradientBoostingRegressor

gbr = GradientBoostingRegressor(n_estimators=100)

# Training the model using cross-validation and calculating the accuracy scores
gbr_scores = cross_val_score(gbr, x_scaled, prices, cv=kf, scoring='neg_mean_absolute_error')

# -----------------------------------------------------------------------
# results of the Gradient Boosting Regressor will be displayed below
# -----------------------------------------------------------------------

print("="*80)
print("RESULTS FOR GRADIENT BOOSTING REGRESSOR MODEL  (negative mean absolute error) ")
print("="*80)

print(f" ::>> ALL Cross-validation scores: \n{gbr_scores}")
print(" ::>> MSE:  %.2f" % gbr_scores.mean())
print(" ::>> Standard deviation: %.2f" % gbr_scores.std())
```

#####  GridSearchCV TECHNIQUE

```python
# Preparing the model for training and testing
# MODEL: GridSearchCV, which is a technique for hyperparameter tuning that performs an exhaustive search over a specified parameter grid. It is used to find the best combination of hyperparameters for a given model by evaluating its performance using cross-validation.

# ----------------------------------------------------------------


from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_squared_error, make_scorer

parameters = {
    'alpha': [0.9, 0.85],
    'learning_rate': [0.01, 0.05, 0.01],
    'max_depth': [2, 3],
    'min_samples_leaf': [5, 3],
    'min_samples_split': [2, 3],
    'n_estimators': [100, 80, 120]
}
gscv =  GridSearchCV(
    estimator=GradientBoostingRegressor(), 
    scoring=make_scorer(mean_squared_error, greater_is_better=False),
    param_grid=parameters, cv=10, n_jobs=-1
)

gscv.fit(x_scaled, prices)

# -----------------------------------------------------------------------
# Results of the GridSearchCV will be displayed below
# -----------------------------------------------------------------------

print("="*85)
print("RESULTS FOR GRIDSEARCHCV (negative mean squared error) across all searched params")
print("="*85)

print(f"::>> The best ESTIMATOR : \n{gscv.best_estimator_}")
print("::>> The best SCORE IS: %.2f" % gscv.best_score_)
print(f"::>> The best PARAMETERS : \n{gscv.best_params_}")
```

#### TRAIN AND TEST GradientBoostingRegressor with GridSearchCV Techmique results

```python
# APLICAR OS MELHORES VALORES PARA os parâmetros dO MODELO DE REGRESSÃO GRADIENT BOOSTING REGRESSOR, COM OS MELHORES PARÂMETROS ENCONTRADOS PELO GRIDSEARCHCV, PARA TREINAR E TESTAR O MODELO COM OS DADOS DE TREINAMENTO E TESTE.

gscv_bp = gscv.best_params_

# gscvbp
# list(gscv_best_params.items())

gbr = GradientBoostingRegressor(
    alpha = gscv_bp['alpha'],
    learning_rate =  gscv_bp['learning_rate'],
    max_depth = gscv_bp['max_depth'],
    min_samples_leaf = gscv_bp['min_samples_leaf'],
    min_samples_split = gscv_bp['min_samples_split'],
    n_estimators = gscv_bp['n_estimators'],
    random_state=30
)
# Training the model using cross-validation and calculating the accuracy scores
gbr_gscv_scores = cross_val_score(gbr, x_scaled, prices, cv=kf, scoring='neg_mean_absolute_error')

# -----------------------------------------------------------------------
# results of the Gradient Boosting Regressor will be displayed below
# -----------------------------------------------------------------------

print("="*80)
print("RESULTS FOR GRADIENT BOOSTING REGRESSOR MODEL V2 (negative mean absolute error) ")
print("="*80)

print(f" ::>> ALL Cross-validation scores: \n{gbr_gscv_scores}")
print(" ::>> MSE:  %.2f" % gbr_gscv_scores.mean())
print(" ::>> Standard deviation: %.2f" % gbr_gscv_scores.std())
```

#### Consolidation of Results

```python
# Consolidação de resultados de todos os modelos testados acima, mostrando os melhores resultados de cada modelo.

# 1 - linear regression model
# 2 - random forest regressor model
# 3 - gradient boosting regressor model
# 4 - gradient boosting regressor model with best parmas estimated by gridsearchcv 

# -----------------------------------------------------------

print("="*80)
print("CONSOLIDATED RESULTS OF ALL MODELS TESTED")
print("="*80)

print("\n")

print("LinearRegressor model\n")
print(" ::>> MSE:  %.2f" % lr_scores.mean())
print(" ::>> Standard deviation: %.2f" % lr_scores.std())
print("-"*80)
print("RandomForestRegressor model\n")
print(" ::>> MSE:  %.2f" % rf_scores.mean())
print(" ::>> Standard deviation: %.2f" % rf_scores.std())
print("-"*80)
print("GradientBoostingRegressor model\n")
print(" ::>> MSE:  %.2f" % gbr_scores.mean())
print(" ::>> Standard deviation: %.2f" % gbr_scores.std())
print("-"*80)
print("GradientBoostingRegressor model with GridSearchCV best params\n")
print(" ::>> MSE:  %.2f" % gbr_gscv_scores.mean())
print(" ::>> Standard deviation: %.2f" % gbr_gscv_scores.std())
print("\n")
print("$"*80) 
```
